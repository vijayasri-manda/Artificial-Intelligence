```python
import pandas as pd
import numpy as np
data = pd.DataFrame({
'Name': ['John','Jane','Bob','Alice'],
'Age': [25, np.nan, 30, 28],
'Score': [85, 90, np.nan, 95]
})
print("Original Data:")
print(data)
```

```python
print("\n Missing Values Count:")
print(data.isnull().sum())
```

```python
data_dropped = data.dropna()
print("\n Data After Dropping Rows with Missing Values:")
print(data_dropped)
```

```python
data_mean_filled = data.copy()
data_mean_filled['Age'] = data['Age'].fillna(data['Age'].mean())
data_mean_filled['Score'] = data['Score'].fillna(data['Score'].mean())

print("\n Data After Filling Missing Values with Mean:")
print(data_mean_filled)
```

```python
data_median_filled = data.copy()

# Fill numeric columns with median
data_median_filled['Age'] = data['Age'].fillna(data['Age'].median())
data_median_filled['Score'] = data['Score'].fillna(data['Score'].median())
print("\n After Filling with Median:")
print(data_median_filled)
```

```python
data_mode_filled = data.copy()
# Fill with mode (most frequent value)
data_mode_filled['Age'] = data['Age'].fillna(data['Age'].mode()[0])
data_mode_filled['Score'] = data['Score'].fillna(data['Score'].mode()[0])

print("\n After Filling with Mode:")
print(data_mode_filled)
```

```python
import pandas as pd
# Create a sample DataFrame
data = {
'A': [1, 2, 3, 4, 5, 100] # 100 is an outlier
}
df = pd.DataFrame(data)
# Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = df['A'].quantile(0.25)
Q3 = df['A'].quantile(0.75)
IQR = Q3 - Q1
# Define bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Eliminate outliers
df_no_outliers = df[(df['A'] >= lower_bound) & (df['A'] <= upper_bound)]
print("DataFrame after removing outliers:\n", df_no_outliers)
```

<h1> Data Preprocessing on the Diabetes Dataset </h1>

```python
from sklearn.datasets import load_diabetes
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif

# Load the Diabetes dataset
diabetes = load_diabetes()

# Convert to DataFrame for better visualization
X = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
y = pd.Series(diabetes.target, name='target')

# Display the first few rows of the data
print("Original Data:")
print(X.head())

# -------------------------------
# 1. Check for missing values
print("\nMissing Values:")
print(X.isnull().sum())

# If there were missing values (just for demonstration purposes)
# You can introduce missing values as follows and fill with mean
# X.iloc[0, 0] = np.nan
# X.iloc[2, 2] = np.nan
X = X.fillna(X.mean())  # Filling missing values with the column mean

# -------------------------------
# 2. Feature scaling
# Min-Max Scaling (0 to 1)
scaler_minmax = MinMaxScaler()
X_minmax = pd.DataFrame(scaler_minmax.fit_transform(X), columns=X.columns)

# Standardization (Z-score)
scaler_standard = StandardScaler()
X_standard = pd.DataFrame(scaler_standard.fit_transform(X), columns=X.columns)

print("\nAfter Min-Max Scaling (first 5 rows):")
print(X_minmax.head())

print("\nAfter Standardization (first 5 rows):")
print(X_standard.head())

# -------------------------------
# 3. Feature selection (Select top 2 features using ANOVA F-test)
selector = SelectKBest(score_func=f_classif, k=2)
X_selected = selector.fit_transform(X, y)
selected_features = X.columns[selector.get_support()]

print("\nSelected top 2 features using ANOVA F-test:")
print(selected_features)

# -------------------------------
# 4. Train-test split (70% training, 30% testing)
X_train, X_test, y_train, y_test = train_test_split(X_selected, y, test_size=0.3, random_state=42)
print("\nTrain and Test Sizes:")
print("Train size:", len(X_train))
print("Test size:", len(X_test))

# -------------------------------
# 5. Binning (discretize a numeric column like 'age')
# Note: Binning works better for categorical data, so you can bin numeric features like 'age'
X['age_bin'] = pd.cut(X['age'], bins=3, labels=["Young", "Middle-aged", "Old"])

print("\nBinned 'age' into categories:")
print(X[['age', 'age_bin']].head())

# -------------------------------
# 6. Summary
print("\nFinal Preprocessed Dataset (first 5 rows):")
print(X.head())
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder
from sklearn.feature_selection import SelectKBest, f_classif

# Load the heart disease dataset
df = pd.read_csv("/content/advertising.tab", delimiter="\t")

# Display first 5 rows of the dataset
print("Original Data:")
print(df.head())

# -------------------------------
# 1. Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Replace non-numeric entries (e.g., 'continuous', 'class') with NaN
df = df.apply(pd.to_numeric, errors='coerce')

# After replacing with NaN, fill missing values:
numeric_columns = df.select_dtypes(include=[np.number]).columns
df[numeric_columns] = df[numeric_columns].fillna(df[numeric_columns].mean())

# For non-numeric columns (like 'sales' or 'TV'), fill with the mode (most frequent value)
non_numeric_columns = df.select_dtypes(exclude=[np.number]).columns
for col in non_numeric_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Check column types again after filling missing values
print("\nData Types After Filling Missing Values:")
print(df.dtypes)

# -------------------------------
# 2. Feature scaling
# Ensure all numeric columns are truly numeric (float)
numeric_columns = df.select_dtypes(include=[np.number]).columns
print("\nValid Numeric Columns After Conversion:")
print(numeric_columns)

# Min-Max Scaling (0 to 1)
scaler_minmax = MinMaxScaler()
df_minmax = df.copy()
df_minmax[numeric_columns] = scaler_minmax.fit_transform(df_minmax[numeric_columns])

# Standardization (Z-score)
scaler_standard = StandardScaler()
df_standard = df.copy()
df_standard[numeric_columns] = scaler_standard.fit_transform(df_standard[numeric_columns])

print("\nAfter Min-Max Scaling (first 5 rows):")
print(df_minmax.head())

print("\nAfter Standardization (first 5 rows):")
print(df_standard.head())

# -------------------------------
# 3. Encode target labels (if the target is categorical)
# For example, let's assume the target column is 'sales' (if applicable)
# Since it's continuous, you may not need to encode, but if you had categorical target, you could do:
# encoder = LabelEncoder()
# df['target'] = encoder.fit_transform(df['target'])

# -------------------------------
# 4. Feature selection (Select top 2 features using ANOVA F-test)
# Assuming the target column is 'sales' (adjust based on your dataset)
X = df.drop(columns=['sales'])  # Remove target column (adjust as needed)
y = df['sales']  # Target column (adjust as needed)

selector = SelectKBest(score_func=f_classif, k=2)
X_selected = selector.fit_transform(X, y)
selected_features = X.columns[selector.get_support()]

print("\nSelected top 2 features using ANOVA F-test:")
print(selected_features)

# -------------------------------
# 5. Train-test split (70% training, 30% testing)
X_train, X_test, y_train, y_test = train_test_split(X_selected, y, test_size=0.3, random_state=42)
print("\nTrain and Test Sizes:")
print("Train size:", len(X_train))
print("Test size:", len(X_test))

# -------------------------------
# 6. Binning (discretize a numeric column)
# Example: We'll bin the 'TV' column into 3 categories
df['TV_bin'] = pd.cut(df['TV'], bins=3, labels=["Low", "Medium", "High"])

print("\nBinned 'TV' into categories:")
print(df[['TV', 'TV_bin']].head())

# -------------------------------
# 7. Summary
print("\nFinal Preprocessed Dataset (first 5 rows):")
print(df.head())
```
