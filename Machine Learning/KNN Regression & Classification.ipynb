**KNN Regression**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsRegressor
```

```python
# Dataset
data = {
'Area': [850, 1000, 1100, 1200, 1300, 1450, 1500, 1600, 1750, 1900, 2000, 2200, 2400, 2500, 2700],
'Bedrooms': [2, 2, 3, 2, 3, 3, 3, 4, 3, 4, 4, 4, 4, 4, 5],
'Price': [15, 18, 20, 22, 24, 26, 29, 31, 34, 37, 39, 43, 46, 48, 52]
}
df = pd.DataFrame(data)
```

```python
# Train model
X = df[['Area', 'Bedrooms']]
Y = df['Price']
knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(X, Y)
```

```python
# New prediction
new_house = pd.DataFrame({'Area': [2100], 'Bedrooms': [4]})
predicted_price = knn.predict(new_house)
```

```python
# Print the predicted value
print("Predicted Price:", predicted_price[0])
```

```python
# Plot
plt.scatter(df['Area'], Y)
plt.scatter(new_house['Area'], predicted_price, marker='x')
plt.xlabel("Area")
plt.ylabel("Price")
plt.show()
```

**KNN Classification**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
# Dataset from image
data = pd.DataFrame({
'BMI': [33.6, 26.6, 23.4, 43.1, 35.3, 35.9, 36.7, 25.7, 23.3, 31],
'Age': [50, 30, 40, 67, 23, 67, 45, 46, 29, 56],
'Sugar': [1, 0, 0, 0, 1, 1, 1, 0, 0, 1]
})
# Split data into features and target
X = data[['BMI', 'Age']]
y = data['Sugar']
# Train the KNN model
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)
# New data to predict
new_person_df = pd.DataFrame([[43.6, 40]], columns=['BMI', 'Age'])
prediction = knn.predict(new_person_df)
# Output the result
print("Prediction", prediction[0])
```
