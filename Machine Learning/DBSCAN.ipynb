**DBSCAN**

```python
 from sklearn.datasets import make_blobs
 from sklearn.cluster import DBSCAN
 import matplotlib.pyplot as plt
 # Create sample data
 X, _ = make_blobs(n_samples=200, centers=3, cluster_std=0.6, random_state=42)
 # Apply DBSCAN
 dbscan = DBSCAN(eps=0.5, min_samples=5)
 labels = dbscan.fit_predict(X)
 # Plot results
 plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='plasma')
 plt.title("DBSCAN Clustering")
 plt.xlabel("Feature 1")
 plt.ylabel("Feature 2")
 plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import pandas as pd
  # Load the Iris dataset
iris = load_iris()
X = iris.data
  # Standardize the features (important for DBSCAN)
X_scaled = StandardScaler().fit_transform(X)
  # Apply DBSCAN
dbscan = DBSCAN(eps=0.6, min_samples=5)
labels = dbscan.fit_predict(X_scaled)
  # Convert to DataFrame for clarity
df = pd.DataFrame(X, columns=iris.feature_names)
df['Cluster'] = labels
  # Display first few rows
print(df.head())
  # Plot the clusters using first two features
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='plasma')
plt.title("DBSCAN Clustering on Iris Dataset")
plt.xlabel("Sepal Length (scaled)")
plt.ylabel("Sepal Width (scaled)")
plt.show()
```
